```python
df['Promo Code'] = df.groupby('Discount Applied')['Promo Code'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))
```

```python
labels = ['young_Adult', 'Adult', 'Middle_aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
df[['age', 'age_group']].head(10)
```

```python
#Create new coloumn purchase_frequency_days

frequency_mapping = {
    'Monthly': 30,
    'Bi-Weekly': 14,
    'Fortnightly': 14,         # Added
    'Weekly': 7,
    'Quarterly': 90,
    'Every 3 Months': 90,      # Added
    'Every 6 Months': 180,
    'Annually': 365            # Added
}

# Re-run the mapping operation
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
import urllib.parse

# Look directly inside your Windows Downloads folder
# Using the 'r' prefix handles Windows folder backslashes correctly
import pandas as pd
df=pd.read_csv('Customer_Shopping _Behavior.csv')

# Step 1: Connect to PostgreSQL
username = "postgres"        
password = "Srinivas@123"    
host = "localhost"          
port = "5433"               
database = "Customer_Shopping_Behavior"  

# Encode the username and password variables
safe_username = urllib.parse.quote_plus(username)
safe_password = urllib.parse.quote_plus(password)

# Create the engine using the safe encoded variables
engine = create_engine(f"postgresql+psycopg2://{safe_username}:{safe_password}@{host}:{port}/{database}")

# Step 2: Load DataFrame into PostgreSQL
table_name = "customer"
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
